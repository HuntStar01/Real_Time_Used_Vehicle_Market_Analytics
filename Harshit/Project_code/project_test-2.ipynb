```python
df_raw = spark.read \
    .option("header", "true") \
    .option("inferSchema", "false") \
    .option("multiLine", "true") \
    .option("quote", '"') \
    .option("escape", '"') \
    .option("mode", "PERMISSIVE") \
    .csv('/user/huntstar/Project/vehicles.csv')
```

```python
# 1. How many rows remain?
print(f"Row count with multiLine: {df_raw.count()}")

# 2. How many distinct state values now?
print(f"Distinct states: {df_raw.select('state').distinct().count()}")
df_raw.groupBy("state").count().orderBy(F.desc("count")).show(20, truncate=False)
```

```python
from pyspark.sql import functions as F

# 1. Exact Count (Slower on massive datasets)
exact_unique_count = df_raw.select(F.count_distinct("posting_date")).collect()[0][0]
print(f"Exact unique values: {exact_unique_count}")

# 2. Approximate Count (Much faster, highly accurate for large data)
approx_unique_count = df_raw.select(F.approx_count_distinct("posting_date")).collect()[0][0]
print(f"Approximate unique values: {approx_unique_count}")
```

```python
from pyspark.sql import functions as F

# 1. Ensure the column is treated as a timestamp, then extract Year and Month
df_with_date = df_raw.withColumn("timestamp_col", F.to_timestamp("posting_date")) \
                 .withColumn("year", F.year("timestamp_col")) \
                 .withColumn("month", F.month("timestamp_col"))

# 2. Get the exact count of unique month-year combinations
unique_months_years_count = df_with_date.select("year", "month").distinct().count()
print(f"Total unique month-year combinations: {unique_months_years_count}")

# 3. View the actual list of unique months and years
df_with_date.select("year", "month").distinct().orderBy("year", "month").show()
```

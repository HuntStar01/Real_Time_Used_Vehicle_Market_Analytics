```python
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .appName("Gold_To_CSV_Export")
    .enableHiveSupport()
    .getOrCreate()
)
```

```python
gold_base = "/user/vehicle_market/gold"

csv_base = "/user/vehicle_market/tableau_csv"
```

```python
gold_tables = [
    "kpi_summary",
    "manufacturer_summary",
    "state_summary",
    "monthly_summary",
    "fuel_summary",
    "segment_summary",
    "condition_summary",
    "title_summary",
    "transmission_summary",
    "state_manufacturer",
    "manufacturer_condition"
]
```

```python
for table in gold_tables:

    print(f"Converting {table}...")

    df = spark.read.parquet(f"{gold_base}/{table}")

    (
        df.write
        .mode("overwrite")
        .option("header", True)
        .csv(f"{csv_base}/{table}")
    )

print("Done!")
```
